```python
df_station = pd.read_csv("data/stations.csv")
```

```python
df_station['last_reported'] = pd.to_datetime(df_station['last_reported'], unit='s').dt.tz_localize(tz="UTC")
df_station['last_reported'] = pd.to_datetime(df_station['last_reported'], unit='s').dt.tz_convert(tz="America/Toronto")
```

```python
df_station.info()
```

```python
df_station['station_id'].values
```

```python
gdf = gpd.GeoDataFrame(
    df_station,
    geometry= gpd.points_from_xy(df_station['lat'], df_station['lon']), # points_from_xy() is an enhanced wrapper for [Point(x, y) for x, y in zip(df.Longitude, df.Latitude)]
    crs="EPSG:4326" # The Coordinate Reference System (CRS)
)

gdf['time'] = gdf['last_reported'].dt.strftime('%Y-%m-%d %I:%M %p')
```

```python
gdf.info()
```

```python
m = folium.Map(location=[gdf['lat'][0], gdf['lon'][0]],
               zoom_start=11,
               tiles="OpenStreetMap")
# create a list of [lat, long] co-ords
points_list = [[points.xy[0][0], points.xy[1][0]] for points in gdf['geometry']]

for i, coords in enumerate(points_list):
    m.add_child(
        folium.Marker(
            location=coords,
            tooltip=gdf['name'][i]
            + "<br>"
            + "station_id: "
            + str(gdf.station_id[i]),
            
            popup="physical_configuration: "
            + str(gdf.physical_configuration[i])
            + "<br>"
            + "capacity: "
            + str(gdf.capacity[i])
            + "<br>"
            + "CREDITCARD: "
            + str(gdf.CREDITCARD[i])
            + "<br>"
            + "num_bikes_available: "
            + str(gdf.num_bikes_available[i])
            + "<br>"
            + "num_docks_available: "
            + str(gdf.num_docks_available[i])
            + "<br>"
            + "is_charging_station: "
            + str(gdf.is_charging_station_y[i])
            + "<br>"
            + "status: "
            + str(gdf.status[i])
            + "<br>"
            + "is_returning: "
            + str(gdf.is_returning[i])
            + "<br>"
            + "last_reported: "
            + str(gdf.time[i])
        )
    )
```
