```python
df = pd.read_excel('../data/2023-WardProfiles-2011-2021-CensusData.xlsx',engine='openpyxl',skiprows=17,usecols="A:AA")
```

```python
df = df.dropna()
```

```python
df.columns
```

```python
col_rename = {}
for col in df.columns:
    if "Ward" in col:
        col_rename[col] = col.split(sep=" ")[1]
```

```python
df = df.rename(columns=col_rename)
```

```python
df['Variables'] = df['Unnamed: 0']
df = df.drop(columns='Unnamed: 0')
```

```python
df = df.set_index('Variables')
```

```python
df.head()
```

```python
df_t = df.T
```

```python
cols = df_t.columns.to_list()
with open("../output/ward_variables.txt",mode='w') as f:
    for x in cols:
        f.writelines(str.strip(x)+"\n")
```
