```python
import pandas as pd
```

# 1. Import data:

```python
df = pd.read_csv('data/toronto-business-licenses-data.csv',encoding='utf-8')
```

# 2. Data Cleaning

```python
df.head()
```

## 2.1. lowercase column names

```python
df.columns
```

```python
df.columns = [c.lower() for c in df.columns]
```

## 2.2. Set correct dtypes
1. datetime: `Issued`, `Cancel Date`, `Last Record Update`
2. Category: `category`, `ward`

### 2.2.1. datetime

```python
df.info()
```

```python
date_cols = ['issued', 'cancel date', 'last record update']
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors = 'coerce')
```

### 2.2.2. Category

```python
df[['category', 'ward']] = df[['category', 'ward']].astype('category')
```
