# Precipitation Analysis 2024
Exploratory Data Analysis and Time Series Analysis of Toronto rainfall data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load Data

```python
df = pd.read_csv('data/precipitation-data-2024.csv')
df['date'] = pd.to_datetime(df['date'])
df.head()
```

## 2. Exploratory Data Analysis (EDA)

```python
df.info()
```

```python
df.describe()
```

### Rainfall Distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['rainfall'], bins=50, kde=True)
plt.title('Distribution of Rainfall Measurements')
plt.xlabel('Rainfall (mm)')
plt.show()
```

### Station Analysis

```python
# Total rainfall per station (approximation using sum of values - note: might need diff logic as per analyze_rain.py, but starting simple)
# Based on analyze_rain.py, simply summing might be wrong if data is cumulative. 
# Let's look at the max value per station as a proxy for cumulative total if it doesn't reset.
station_stats = df.groupby('name')['rainfall'].agg(['max', 'sum', 'count']).reset_index()
station_stats.sort_values('max', ascending=False).head(10)
```

### Geospatial Visualization

```python
plt.figure(figsize=(10, 8))
sns.scatterplot(data=df.drop_duplicates(subset=['name']), x='longitude', y='latitude', hue='name', legend=False)
plt.title('Station Locations')
plt.show()
```

## 3. Time Series Analysis

```python
# Resample to daily frequency
daily_rain = df.set_index('date').resample('D')['rainfall'].sum().reset_index()

plt.figure(figsize=(14, 6))
sns.lineplot(data=daily_rain, x='date', y='rainfall')
plt.title('Total Daily Rainfall in Toronto (2024)')
plt.xlabel('Date')
plt.ylabel('Total Rainfall (mm)')
plt.show()
```
