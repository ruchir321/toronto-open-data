# Toronto Signature sites

Signature Sites is a listing of select industrial and commercial development and redevelopment properties for sale or lease in Toronto

[Link](https://www.toronto.ca/business-economy/business-operation-growth/signature-sites-toronto-site-selection-tool/#location=&lat=&lng=&zoom=)

```python
import geopandas as gpd
import folium
import pandas as pd
import json
from shapely.geometry import shape
```

```python
pd.set_option("display.max_columns", None)
```

```python
df = gpd.read_file("../data/resource_dump_0.csv")
```

```python
def parse_geometry(geom_str):
    if pd.isna(geom_str):
        return None
    try:
        geom_dict = json.loads(geom_str)
        return shape(geom_dict)
    except (json.JSONDecodeError, TypeError):
        return None
```

```python
df['geometry'] = df['geometry'].apply(parse_geometry)
```

```python
gdf = gpd.GeoDataFrame(df, geometry="geometry")
gdf.head()
```

```python
gdf.info()
```

```python
m = folium.Map(location=[43.6532, -79.3832], zoom_start=12)

geom_list = [[point.xy[1][0], point.xy[0][0]] for point in gdf.geometry]

for i, coords in enumerate(geom_list):
    m.add_child(
        folium.Marker(
            popup="Site Name: "
            + str(gdf.site_name[i]),
            location=coords,
            icon=folium.Icon()
        )
    )

m
```
